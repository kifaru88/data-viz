```python
import pandas as pd
import numpy as np
```

```python
np.random.seed(42)
```

```python
n = 300
```

```python
harvest_kg = np.random.uniform(20 , 1000 , n)
rainfall_mm = np.random.uniform( 10 , 200 , n)
```

```python
print(len(harvest_kg))
print(len(rainfall_mm))
```

```python
#simulated honey market r/ship -- polynomial terms
price = (
    650 #starting point
    - 0.18 * harvest_kg # Linear harvest effect
    + 0.00015 * harvest_kg ** 2 #Quadratic harvest effect
    + 0.9 * rainfall_mm
    - 0.002 *rainfall_mm ** 2
    + 0.0003 * harvest_kg * rainfall_mm
    + np.random.normal ( 0, 35, n)
)
```

```python
df = pd.DataFrame({"harvest_kg":harvest_kg,
                 "rainfall_mm":rainfall_mm,
                 "price_ksh_per_kg":price})
```
