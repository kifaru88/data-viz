```python
import pandas as pd
import numpy as np
```

```python
Data = {
    "Cumstomer" : ["c001","c002","c003","c004","c005","c006","c007","c008"] ,
    "Distance_km" : [2,15,7,25,4,12,30,6] ,
    "Delivery_Time_min" : [25,70,40,110,30,65,140,35] ,
    "Orders_per_month" : [12,4,8,2,15,6,1,10] ,
    "Money_spent_KES" : [8400,5200,11000,3500,14500,7600,2200,12500] 
}
```

```python
df = pd.DataFrame(Data)
```

```python
df.describe()
```

```python
df.head()
```

```python
#Money_spent_KES has the highest value
#Orders_per_month has the smallest 
#Yes the difference could affect K-Means
```

```python
Distance_normalized =( df["Distance_km"] - df["Distance_km"].min() ) / (df["Distance_km"].max() -df["Distance_km"].min())

print(Distance_normalized)
```

```python
Time_normalized = (df["Delivery_Time_min"] - df["Delivery_Time_min"].min()) / (df["Delivery_Time_min"].max() -df["Delivery_Time_min"].min())
print(Time_normalized)
```

```python
Orders_normalized = (df["Orders_per_month"] - df["Orders_per_month"].min()) / (df["Orders_per_month"].max() -df["Orders_per_month"].min())
print(Orders_normalized)
```

```python
Money_normalized = (df["Money_spent_KES"] - df["Money_spent_KES"].min()) / (df["Money_spent_KES"].max() -df["Money_spent_KES"].min())
print(Money_normalized)
```
