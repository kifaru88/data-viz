```python
missing_data = df[["age","employer_name","monthly_contribution","phone_number","remarks"]].isnull().sum()
print(missing_data)
```

```python
original_data = len(df)
print(original_data)
```

```python
#finding the percentage of each column missing
percentage_missing = missing_data / original_data *100
print(percentage_missing.round(2).astype(str) + "%")
```

```python
#Cleaning age column
df["age"] = (df["age"].astype("string")
    .str.replace("yrs","",regex=False)
    .str.replace("years","",regex=False))
```

```python
df["age"] = pd.to_numeric(df["age"],errors="coerce")
```
