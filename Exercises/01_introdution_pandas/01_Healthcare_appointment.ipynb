```python
import pandas as pd
import numpy as np

data = {
    "patient_id": [101, 102, 103, 104, 105, 106, 107, 108, 109, 110],
    "age": [25, 34, "?", 67, 45, 29, 52, "?", 38, 61],
    "appointment_date": [
        "2026-08-01",
        "2026-08-01",
        "?",
        "2026-08-02",
        "2026-08-02",
        "?",
        "2026-08-03",
        "2026-08-03",
        "2026-08-04",
        "?"
    ],
    "department": [
        "Dental",
        "Cardiology",
        "General",
        "?",
        "Dental",
        "General",
        "Cardiology",
        "?",
        "Dental",
        "General"
    ],
    "appointment_status": [
        "Attended",
        "Missed",
        "Attended",
        "?",
        "Attended",
        "Missed",
        "Attended",
        "Attended",
        "?",
        "Missed"
    ]
}

df = pd.DataFrame(data)
```

```python
#Display the first 5 rows
df.head(5)
```

```python
#Finding the number of rows and columns
df.shape
```

```python
df.dtypes
#age is not numerical as it contains '?' ,thus pandas treats it as an object
```

```python
#identifing missing data
df_clean = df.replace("?",np.nan)
```

```python
#count missing values in each column
df_clean.isna().sum()
#appointment_date is the column with the most missing data
#patient_id has no missing data
```

```python
#Removing records where appointment data is missing
df = df_clean.dropna(subset=["appointment_date"] , axis=0)
```

```python
#number of records before cleaning is
Original_records = len(df_clean)
print(f"len:{Original_records:.2f}")
```

```python
#number of records after cleaning is 
Cleaned_records = len(df)
print(f"len:{Cleaned_records:.2f}")
```

```python
#number of records removed is 
Records_removed = Original_records - Cleaned_records
print(f"Records removed:{Records_removed}")
```

```python
#Percentage of records removed
Percentage_removed = ( Records_removed / Original_records) *100
```
